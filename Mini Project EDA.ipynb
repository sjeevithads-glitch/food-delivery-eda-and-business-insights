### 📊 Findings
1. The **slowest average delivery area** is **OMR**, with mean delivery time of **41.2 minutes** and some extreme delays up to **240 minutes**.  
2. The **cuisine with the longest kitchen time** is **Biryani**, averaging **19 minutes** of preparation.  
3. The **probability of delivery exceeding 45 minutes** is **18.3%**, showing nearly one in five orders face significant delays.  
   - Outliers include **order_value = ₹12,800** and **rating = 50.0**, which are likely data errors.

### ✅ Recommendations
- **Prioritize complaint handling and high‑value orders**: Customers with unusually high bills or complaints are at greater churn risk; proactive support can reduce losses.  
- **Streamline Biryani preparation**: Since it consistently takes longer, invest in process improvements or pre‑prep strategies to cut kitchen time.

### ⚠️ Limitation
This dataset cannot explain **why customers rate orders poorly** — ratings are influenced by taste, service, and packaging, none of which are captured in the numeric fields. Qualitative feedback is needed to understand customer sentiment.
